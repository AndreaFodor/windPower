# Baseline model

In this notebook we will run and test a dummy model on our data.

```python
import pandas as pd
import copy
import numpy as np
```

```python
traindf = pd.read_csv('../../1_data/final_dataframes/main_validation_dataframe.csv')

testdf = pd.read_csv('../../1_data/final_dataframes/main_testing_dataframe.csv')
testdf = testdf.iloc[35064:43824]
```

```python
#testing the dummy regressor
#findme

#now we need to take the entries in 'Wind_power' and then add n=24 of them together for power one day ahead
n=24
traindf['Wind'] = traindf['Wind'].rolling(window=n).sum().shift(-n) #this replaces 'Wind' time t with the sum of the entries in 'Wind' at t+1,... t+24
traindf = traindf.iloc[:-24] #removing last 24 entries
traindf = traindf[::24]
#now we need to take the entries in 'Wind_power' and then add n=24 of them together for power one day ahead
n=24
testdf['Wind'] = testdf['Wind'].rolling(window=n).sum().shift(-n) #this replaces 'Wind' time t with the sum of the entries in 'Wind' at t+1,... t+24
testdf = testdf.iloc[:-24] #removing last 24 entries
testdf = testdf[::24]
```

# Testing the dummy on first 6 months of 2022

```python
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_percentage_error, mean_absolute_error, r2_score
```

```python
df =traindf.iloc[0:1096 ]
X_train = df[[col for col in df.columns if col != "Wind"]]
y_train = df["Wind"]

df2 =traindf.iloc[1096: 1277]
X_test = df2[[col for col in df2.columns if col != "Wind"]]
y_test = df2["Wind"]

#dummy regressor
dummy = DummyRegressor(strategy='mean')  # You can also use strategy='constant', constant=YOUR_VALUE
dummy.fit(X_train, y_train)

# Predictions
y_pred = dummy.predict(X_test)

# Evaluation
print("MAPE:", mean_absolute_percentage_error(y_test, y_pred))
print("R^2 Score:", r2_score(y_test, y_pred))
print("MAE:", mean_absolute_error(y_test, y_pred))
print("Mean", float(y_pred[0:1]))
```

```python
df =traindf
X_train = df[[col for col in df.columns if col != "Wind"]]
y_train = df["Wind"]

df2 =testdf
X_test = df2[[col for col in df2.columns if col != "Wind"]]
y_test = df2["Wind"]

#dummy regressor
dummy = DummyRegressor(strategy='mean')  # You can also use strategy='constant', constant=YOUR_VALUE
dummy.fit(X_train, y_train)

# Predictions
y_pred = dummy.predict(X_test)

# Evaluation
print("MAPE:", mean_absolute_percentage_error(y_test, y_pred))
print("R^2 Score:", r2_score(y_test, y_pred))
print("MAE:", mean_absolute_error(y_test, y_pred))
print("Mean", float(y_pred[0:1]))
```
